```python
import pandas as pd
```

```python
df=pd.read_csv(r'Downloads\imdb_movies.csv')
df
```

```python
pd.set_option('display.max.columns',25)
```

```python
df
```

```python
df.drop_duplicates(inplace=True)
```

```python
df
```

```python
df.dropna(subset=['genres'],inplace=True)
```

```python
df.info()
```

```python
df['profit']=df['revenue']-df['budget']
```

```python
df.head()
```

```python
df1=df[['id','popularity','budget','revenue','profit','original_title','runtime','genres','release_year','vote_count','vote_average']]
```

```python
df1
```

```python
df1['genres']=df1['genres'].str.split('|').str[0]
```

```python
df1
```

```python
# Which genres are the most common?
genre_count=df1.groupby('genres')['id'].count().sort_values(ascending=False)
genre_count
genre_count.plot.pie(title='Movies per genre %',autopct='%1.1f%%',figsize=(12,14))
```

```python
genre_count.plot.barh(title='Movies per genre %',figsize=(12,10),color='indigo')
```

```python
# Which genres have a high avg profit?
genre_profit=df1.groupby('genres')['profit'].mean().sort_values(ascending=False)
pd.options.display.float_format='{:2f}'.format
genre_profit

```

```python
genre_profit.plot.barh(title='Profit by Genre',figsize=(12,8),color='darkblue')
```

```python

genre_rev=df1.groupby('genres')[['budget','revenue']].mean().sort_values(by='revenue',ascending=False)
genre_rev.plot.barh(title='Budget vs Revenue by Genre',figsize=(12,8),color=('brown','green'))
```

```python
# Which genre has a high popularity?
genre_rating=df1.groupby('genres')['popularity'].mean().sort_values()
genre_rating
```

```python
genre_rating.plot.barh(title='Genre average popularity', figsize=(13,6))
```

```python
# Which genres have a rating higher than 8
highly_rated=df1[df1['vote_average']>=8].sort_values(by='vote_average')
hr=highly_rated.groupby('genres').count()
```

```python
hr=hr[['id']]
hr.sort_values(by='id',ascending=False,inplace=True)
```

```python
hr.plot.barh(title='Highest Rating count by Genre')
```

```python
# Hypothesis 1: Movies according to vote avg return high profit and revenue
highly_rated.corr(numeric_only=True,method='spearman')
# Hypothesis is rejected. Vote average has a minimal negative correlation with revenues and profits.
# Looking closely at the votecount and popularity correlation with revenues and profits, the data suggests that the number of people watching movies is marginaly more tha the number of people watching the movies and voting.
# Further, vote_count may occur after the movie such that the data on votecount may not have much impact on future revenues hence the lower correlation coeffecient.
```

```python
df1.plot.scatter(x='vote_average',y='revenue')
```

```python
highly_rated.corr(numeric_only=True,method='spearman')
```

```python
# Hypothesis 2: The best movies according to popularity return high profit and revenue
df1.corr(numeric_only=True)
#  Popularity is correlated with revenues and profits; 73% and 66% respectively.
```

```python
# Hypothesis 3: Highly budgeted movies return high revenue and profit.
import matplotlib.pyplot as plt
import seaborn as sns
sns.regplot(x='budget',y='profit',data=df1,line_kws={'color':'black'})

# There is a significant positive correlation to support the hypothesis.
# It is possible highly budgeted movies spend resources marketing the movies to attract a wider audience resulting in higher profit.
```

```python
# Hypothesis 4: Highly budgeted movies have a high popularity
sns.regplot(x='budget',y='popularity',data=df1,line_kws={'color':'black'})
# There is a significant positive correlation to support the hypothesis.
# It is possible highly budgeted movies spend resources marketing the movies to attract a wider audience resulting in higher popularity.
```

```python
time_genre=pd.DataFrame(df1.groupby(['release_year','genres'])['profit'].mean())
time_genre
```

```python
time_genre1=pd.DataFrame(df1.groupby(['release_year','genres'])['profit'].count())
time_genre1.rename(columns={'profit':'count'},inplace=True)
time_genre1
```

```python
evolution=pd.pivot_table(time_genre1,values='count',index=['genres'],columns=['release_year'])
```

```python
sns.heatmap(evolution,cmap='rocket',linewidths=.6)
plt.title('Genres count by Year')
```

```python
final=pd.pivot_table(time_genre,values='profit',index=['genres'],columns=['release_year'])
```

```python
sns.set(rc={'figure.figsize':(15,10)})
sns.heatmap(final,cmap='crest',linewidths=.6)
plt.title('Genres by Profit per Year')
```
